# Bronze Layer - Kaggle E-commerce Data Ingestion

This notebook creates bronze tables from CSV files in the `kaggle_e_com_files` volume.

Source Volume: `resume_project.default.kaggle_e_com_files`  


```sql
-- Bronze Table: Aisles
CREATE OR REPLACE TABLE resume_project.default.bronze_aisles AS
SELECT 
  aisle_id,
  aisle
FROM read_files(
  '/Volumes/resume_project/default/kaggle_e_com_files/aisles.csv',format => 'csv',header => true);
```

```sql
-- Bronze Table: Departments
CREATE OR REPLACE TABLE resume_project.default.bronze_departments AS
SELECT 
  department_id,
  department
FROM read_files(
  '/Volumes/resume_project/default/kaggle_e_com_files/departments.csv',format => 'csv',header => true);
```

```sql
-- Bronze Table: Products
CREATE OR REPLACE TABLE resume_project.default.bronze_products AS
SELECT 
  product_id,
  product_name,
  aisle_id,
  department_id
FROM read_files(
  '/Volumes/resume_project/default/kaggle_e_com_files/products.csv',format => 'csv',header => true);
```

```sql
-- Bronze Table: Orders
CREATE OR REPLACE TABLE resume_project.default.bronze_orders AS
SELECT 
  order_id,
  user_id,
  eval_set,
  order_number,
  order_dow,
  order_hour_of_day,
  days_since_prior_order
FROM read_files('/Volumes/resume_project/default/kaggle_e_com_files/orders.csv',format => 'csv',header => true);
```

```sql
-- Bronze Table: Order Products (Prior Orders)
CREATE OR REPLACE TABLE resume_project.default.bronze_order_products_prior AS
SELECT 
  order_id,
  product_id,
  add_to_cart_order,
  reordered
FROM read_files(
  '/Volumes/resume_project/default/kaggle_e_com_files/order_products__prior.csv',format => 'csv',header => true);
```

```sql
-- Bronze Table: Order Products (Training Set)
CREATE OR REPLACE TABLE resume_project.default.bronze_order_products_train AS
SELECT 
  order_id,
  product_id,
  add_to_cart_order,
  reordered
FROM read_files(
  '/Volumes/resume_project/default/kaggle_e_com_files/order_products__train.csv',format => 'csv',header => true);
```
