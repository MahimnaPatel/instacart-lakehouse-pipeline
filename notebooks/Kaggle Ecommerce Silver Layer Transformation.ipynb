# Silver Layer - Kaggle E-commerce Data Transformation

This notebook transforms bronze tables into cleaned silver tables.

**Source Tables:** `resume_project.default.bronze_*` (bronze layer)  
**Target Schema:** `resume_project.default`  
**Silver Tables:** aisles, departments, products, orders, prodprior, prodtrain

**Data Quality:** Filters nulls, validates data types, removes invalid records

```sql
-- Silver Table: Aisles
CREATE OR REPLACE TABLE resume_project.default.aisles AS
SELECT 
  aisle_id,
  TRIM(aisle) as aisle
FROM resume_project.default.bronze_aisles
WHERE aisle_id IS NOT NULL
  AND aisle IS NOT NULL;
```

```sql
-- Silver Table: Departments
CREATE OR REPLACE TABLE resume_project.default.departments AS
SELECT 
  department_id,
  TRIM(department) as department
FROM resume_project.default.bronze_departments
WHERE department_id IS NOT NULL
  AND department IS NOT NULL;
```

```sql
-- Silver Table: Products
CREATE OR REPLACE TABLE resume_project.default.products AS
SELECT 
  product_id,
  TRIM(product_name) as product_name,
  CAST(aisle_id AS INT) as aisle_id,
  CAST(department_id AS INT) as department_id
FROM resume_project.default.bronze_products
WHERE product_id IS NOT NULL
  AND product_name IS NOT NULL
  AND aisle_id RLIKE '^[0-9]+$'
  AND department_id RLIKE '^[0-9]+$';
```

```sql
-- Silver Table: Orders
CREATE OR REPLACE TABLE resume_project.default.orders AS
SELECT 
  order_id,
  user_id,
  TRIM(eval_set) as eval_set,
  order_number,
  order_dow,
  order_hour_of_day,
  days_since_prior_order
FROM resume_project.default.bronze_orders
WHERE order_id IS NOT NULL
  AND user_id IS NOT NULL
  AND eval_set IS NOT NULL;
  days_since_prior_order IS NOT NULL;
```

```sql
-- Silver Table: Order Products
CREATE OR REPLACE TABLE resume_project.default.prodprior AS
SELECT 
  order_id,
  product_id,
  add_to_cart_order,
  reordered
FROM resume_project.default.bronze_order_products_prior
WHERE order_id IS NOT NULL
  AND product_id IS NOT NULL
  AND add_to_cart_order IS NOT NULL
  AND reordered IS NOT NULL;
```

```sql
-- Silver Table: Order Products
CREATE OR REPLACE TABLE resume_project.default.prodtrain AS
SELECT 
  order_id,
  product_id,
  add_to_cart_order,
  reordered
FROM resume_project.default.bronze_order_products_train
WHERE order_id IS NOT NULL
  AND product_id IS NOT NULL
  AND add_to_cart_order IS NOT NULL
  AND reordered IS NOT NULL;
```
